# KORE Engine — Interactive SQL Demo

This notebook demonstrates the KORE SQL kernel for Jupyter.
It shows loading data, running SQL queries, using magic commands, and mixing SQL with Python.

## 1. Loading Data

Use the `%load_csv` magic command to load a CSV file into a table.

```kore-sql
%load_csv ../tests/fixtures/people.csv AS people
```

## 2. Listing Tables

Use `%tables` to see all registered tables.

```kore-sql
%tables
```

## 3. Inspecting Schema

Use `%schema tablename` to see the table structure.

```kore-sql
%schema people
```

## 4. Running SQL Queries

Type SQL directly — no magic prefix needed. Results are displayed as HTML tables.

```kore-sql
SELECT * FROM people
```

```kore-sql
SELECT name, city FROM people WHERE age > 25 ORDER BY name
```

## 5. Explain Query Plan

Use `%explain` to see how KORE will execute a query.

```kore-sql
%explain SELECT * FROM people WHERE city = 'NY' ORDER BY age
```

## 6. Timed Execution

Use `%time` to measure query performance.

```kore-sql
%time SELECT * FROM people ORDER BY age
```

## 7. Python Code Blocks

Use `%%python` to execute Python code with access to the `spark` session.

```kore-sql
%%python
df = spark.sql("SELECT * FROM people")
rows = df.collect()
print(f"Total rows: {len(rows)}")
for row in rows:
    print(f"  {row['name']} (age {row['age']}) — {row['city']}")
```

## 8. Creating Data Programmatically

You can create tables from Python data structures.

```kore-sql
%%python
data = [
    {"product": "Widget", "sales": "150", "region": "East"},
    {"product": "Gadget", "sales": "200", "region": "West"},
    {"product": "Widget", "sales": "100", "region": "West"},
    {"product": "Gadget", "sales": "175", "region": "East"},
]
df = spark.createDataFrame(data)
print(f"Created DataFrame with {df.count()} rows")
print(f"Columns: {df.columns}")
```

## 9. Visualizing Results

Aggregate queries may generate chart suggestions automatically.

```kore-sql
SELECT city, COUNT(*) as count FROM people GROUP BY city
```

## Summary

The KORE SQL kernel supports:
- **Direct SQL** — type queries directly, get HTML-formatted results
- **Magic commands** — `%tables`, `%schema`, `%load_csv`, `%load_parquet`, `%explain`, `%time`
- **Python blocks** — `%%python` for full programmatic access
- **Auto-visualization** — simple bar charts for aggregate results
- **Tab completion** — table names and magic commands
